```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import kagglehub
path = kagglehub.dataset_download("computingvictor/transactions-fraud-datasets")
```

```python
import os

# List the contents of the directory
print(os.listdir(path))
```

```python
df = pd.read_csv(os.path.join(path, 'cards_data.csv'))

print(f"First 10 rows of the data:\n")
display(df.head(5))
```

```python
df.shape
```

```python
print(df['card_brand'].value_counts())
print(df['card_type'].value_counts())
print(df['has_chip'].value_counts())
print(df['card_on_dark_web'].value_counts())
```

```python
from sklearn.preprocessing import LabelEncoder as label_encoder
le = label_encoder()
df['card_type'] = le.fit_transform(df['card_type'])

print(f"Original Categories: {le.classes_}")
print(f"Encoded Categories: {le.transform(le.classes_)}")
```
